# Checkpoint Transfer

Modeli sıfırdan eğitmek çok zor olduğu için orjinal modelden eğitilmiş ağırlıkları kendi modelimize aktardık. Yani modelin öğrenmiş olduğu  dile/içeriğe bağımsız becerileri (temporal bloklar, cross-modal fusion, attention, head) kendi modelimize taşıyoruz.

Model eğitilirken 2 dataset üzerinde eğilmiş, Mr.Hisum ve MoSu verisetleri. Bu yüzden 2 ayrı checkpointimiz var. Mr.HiSum inceptionv3 kullandığı için onu seçtik.

Bu taşıma işleminde sadece visual_proj da boyut uyuşmazlığı olduğu için sıfırdan eğitildi. Model [128, 1024] boyutlu bir matris kullanırken, biz [128, 2048] boyutunda matris kullanıyoruz. BU sebeple visual_proj kopyalanmadı, sıfırdan başlatıldı.

Ayrıca model orjinalinde text encoder olarak RoBERTa kullanmıştı, biz ise BERTurk kullanıyoruz. Bu mekanik olarak kopyalandı, ama İngilizce RoBERTa dağılımına ait olduğu için Türkçe BERTurk'e fine-tune ile uyarlanacak.

Bunun dışında katmanların çoğu (temporal, fusion, attention, head, audio_proj, embedding'ler) birebir orjinal modelden kendi modelimize taşındı.

Checkpoint transfer: modele başlangıç noktalarını (Mr.HiSum'ın öğrendiği ağırlıklar) yükledik.

bir sonraki aşama olan training ise bu başlangıç noktalarını kendi verinle (gt_score) fine-tune edeceğiz..
